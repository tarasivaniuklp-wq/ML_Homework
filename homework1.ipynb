```python
Asia_cars = df[df['origin'] == 'Asia']
target_columns = Asia_cars[['vehicle_weight', 'model_year']]
target_columns = target_columns.head(n=7)
X = np.array(target_columns)
TX = X.T
XTX = TX.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T) 
w = w.dot(y)
print(w.sum())
```
